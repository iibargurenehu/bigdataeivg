```python
velas = compraventas \
    .withColumn( \
        "bucket_time", \
        date_trunc("hour", col("time")) \
    ) \
    .groupBy( \
        "symbol", \
        "bucket_time" \
    ) \
    .agg( \
        min_by( \
            col("price"), \
            col("time") \
        ).cast("float").alias("open"), \

        max("price") \
            .cast("float") \
            .alias("high"), \

        min("price") \
            .cast("float") \
            .alias("low"), \

        max_by( \
            col("price"), \
            col("time") \
        ).cast("float").alias("close"), \

        sum("qty") \
            .cast("float") \
            .alias("volume") \
    ) \
    .select( \
        col("bucket_time").alias("time"), \
        col("symbol"), \
        lit("1h").alias("resolution"), \
        col("open"), \
        col("high"), \
        col("low"), \
        col("close"), \
        col("volume") \
    ) 
```
